```python
import numpy as np

# ----------------------------------
# arange()
# ----------------------------------
print(np.arange(10))
print(np.arange(1, 20, 2))

# ----------------------------------
# reshape()
# ----------------------------------
arr = np.arange(12)
print(arr.reshape(3, 4))
print(arr.reshape(2, 6))

# ----------------------------------
# random functions
# ----------------------------------
print(np.random.random((3, 3)))

print(np.random.randint(1, 100, 10))

# ----------------------------------
# seed()
# ----------------------------------
np.random.seed(1)
print(np.random.randint(1, 100, 10))

# ----------------------------------
# choice()
# ----------------------------------
print(np.random.choice([10, 20, 30, 40, 50]))

# ----------------------------------
# permutation()
# ----------------------------------
print(np.random.permutation([1, 2, 3, 4, 5]))

# ----------------------------------
# linspace()
# ----------------------------------
print(np.linspace(-10, 10, 10))
print(np.linspace(1, 100, 25))

# ----------------------------------
# sort()
# ----------------------------------
x = np.array([5, 8, 1, 4, 9, 3])
print(np.sort(x))

y = np.array([[5, 2, 8],
              [9, 1, 3]])

print(np.sort(y))

# ----------------------------------
# append()
# ----------------------------------
a = np.arange(6)
print(np.append(a, 200))

b = np.array([[1, 2, 3],
              [4, 5, 6]])

print(np.append(b, [[7, 8, 9]], axis=0))

# ----------------------------------
# concatenate()
# ----------------------------------
c = np.arange(6)
d = np.arange(6, 12)

print(np.concatenate((c, d)))

# ----------------------------------
# unique()
# ----------------------------------
e = np.array([1, 2, 2, 3, 4, 4, 5])
print(np.unique(e))

# ----------------------------------
# expand_dims()
# ----------------------------------
f = np.arange(6)

print(np.expand_dims(f, axis=0))
print(np.expand_dims(f, axis=1))

# ----------------------------------
# where()
# ----------------------------------
g = np.array([10, 20, 30, 40, 50])

print(np.where(g > 25))

print(np.where(g > 25, "yes", "no"))

# ----------------------------------
# argmax() and argmin()
# ----------------------------------
h = np.array([10, 100, 45, 2, 87])

print(np.argmax(h))
print(np.argmin(h))

# ----------------------------------
# cumsum() and cumprod()
# ----------------------------------
i = np.array([1, 2, 3, 4, 5])

print(np.cumsum(i))
print(np.cumprod(i))

# ----------------------------------
# percentile()
# ----------------------------------
j = np.array([10, 20, 30, 40, 50])

print(np.percentile(j, 25))
print(np.percentile(j, 50))
```
